1. Q. What is Anomaly Detection? Explain its types (point, contextual, and
collective anomalies) with examples.

>> A. Anomaly Detection (also known as outlier detection) is the process of identifying rare items, events, or observations that deviate significantly from the majority of the data and do not conform to the expected normal behavior or pattern. These anomalies often indicate critical incidents such as fraud, faults, cyberattacks, system failures , or novel discoveries. It is an unsupervised learning task (most commonly) used in domains like cybersecurity, finacy, manufacturing, healtcare, and network monitoring.

Anomalies are broadly classified into three main types:

- **Point Anomalies** (also called global outliers): These occur when a single data instance is significantly different from the rest of the data. the individual point deviates markedly from the normal distribution or pattern.
**Ex** : In credit card transaction data, a sudden withdrawal of 5,00,000 from a user who typically spends 2.000- 10,000 per transaction is point anomaly (potential fraud). Another example is a temperature sensor in a sensor room suddenly reading 80 C while all other readings are around 25 -30 C.

- **Contextual Anomalies** (also called conditional anomalies): These occur when a data point is anomalous only in a specific context, but it would appear normal in another context. The anomaly depends on contextual attributes (e.g., time, location, season). The value itself may lie within the normal global range.

**Ex** : A website traffic drop of 40% is normal on public holiday (low activity expected), but the same drop on a regular weekday would be a contextual anomaly (indicating possible site outage). Another example: High electricity usage at 3 AM in a household in contextual (family usually sleeps), but the same usage at 9 PM is normal.

- **Collective Anomalies**: These occur when a collection or group of relateed data points is anomalous together, even though each individual point apperas normal when viewed in isolation. The anomaly is in the collective behaviour or pattern formed by the group.
**Ex**: In network traffic monitioring, a sequenece of login attempts from different countries within a few minutes (each attempt normal individually) forms a collective anomaly indicating a possible credential stuffing attack. Another example : In ECG heart monitoring, a short sequence of irregular heartbeats may form a collective anomaly(arrhythmia), even if individual beats are within acceptable ranges.

>> A. Isolation Forest, DBSCAN, and Local Outlier Factor (LOF) are popular unsupervised anomaly /outlier detection algorithms. Here's a comparions:

**Aspect > Isolation Forest(iForest) > DBSCAN  (Density - Based Spatial Clustering of Applications with Noise) > Local Outlier Factor (LOF)**

- **Core Approach** > Uses random partitioning (ensemble of isolation trees) to isolate points. Anomalies are "few and different" - isolated foster (shorter path length in trees). Does not model normal data density explicitly > Density -based clusterin: Groups points in high- density regions into clusters; marks low-density points (not reachable from core points) as noise/outliers > Density-based: Measures local density deviation of a point compared to its K-nearest neighbour. points and significantly lower local density than neighbors get high LOF scores(outliers).

- **Key Mechansim** > Random recursive splitting (axis-parallel); anomaly score = average path length ( shorter = more anomalous) > Defines core points, border points, and noise using e (radius) and Minpts parameters > Compares reachability distance and local reachability density; LOF > 1 indicates outlier.

- **Strengths** > Very efficient and scalable (O (nlog N)); works well in high dimensions; no distance computations needed; handles irrelevant features better > Naturally identifies outliers as noise; handles arbitrary shapes; no need to specify number of clusters > Good at detecting local outliers in datasets with varying densities (non-uniform clusters).

- **Limitations** > Assumes anomalies are isolated (less effective for clustered anomalies or very subtle local outliers); random nature- some variance (mitigated by ensemble) > Sensitive to parameters 0eand Minpts ); struggles in high dimensions ( Curse of dimensionality), not ideal for very large datasets without tuning > Computationally expensive (o(n2) without approximates); sensitive to K parameter; poor high dimensions or uniform density data.

- **Suitable Use Cases** > High- dimesnional data (e.g., fraud detection in transactions, network intrusion); large-scale datasets; quick detection of global/point anomalies > Spatial/geographical data; datasets with clusters of varying shapes and densities (e.g., spatial sensor faults); when natural clusters exist and outliers are noise > Datasets with varying local densities (e.g., manufacturying defect detection where normal regions have different densities ); detecting local/ subtle outliers in non-uniform data.

3. Q.  What are the key components of a Time Series? Explain each with one
example.

>> A. A time series is sequence of data points collected or recorded at successive time intervals (usually equally spaced ). It can be decomposed into four key components that explain the underlying patterns and variations in the data. These components are:

1. **Trend**: The long-term progression or direction in the data over an extended period. It represents a gradual increase, decease, or stability in the level of the series, ignoring short-term fluctuations.
**Ex** : The steady upward increase in global average temperature over the past century (climate change trend), even though there are yearly fluctuations.

2. **Seasonlity**: Regular, predictable, and repeating patterns that occur a fixed intervels due to seasonal factors (e.g., calender-based like months, queaters, or days of the week). the pattern repreats consistently in timing, direction, and  magnitude.
**EX**: Monthly retail sales spiking every december due to hliday shopping and dipping in january, repeating every year.

3. **Cyclical (or Cycle): Longer- terms fluctuations or waves that are not fixed to a calender and do not have a consistent period (usually lasting several years). These are often tied to economic, business, or external cycles and are more irregular than seasonality. **EX**: Business/economic cycles, such as periods of economic expansion (boom) followed by contraction (recession) every 5-10 years, like the global finanical cycles observed in GDP data.

4. **Irregular( or Random/Residual/Noise):** Unpredictable, random variations or shocks that cannot be explained by trends, seasonality, or cycical components. These are short terms anomalies or noise caused by unforeseen events.
**EX**; A sudden drop in airline passenger numbers due to an unexpected event like a volcanic eruption grounding flights or a natural disaster.

These components are often modeled additively (Y= T + S + c + I) or multiplicatively ( Y = T x S x C x I).

4. Q.  Define Stationary in time series. How can you test and transform a
non-stationary series into a stationary one?
>> A. ** Definition of Stationarity:** A time series is stationary if its statistical properties remain constant over time. Specifically, in weak (or covariance) stationarity (the most commonly used form):

- The Mean is constant (no trend or level shifts).

- The variance is constant (homoscedasticity, no changing spread).

- The autocovariance (or autocorrelation) depends only on the lag between obersvations, not on time.

Stricts stationally requires the entire joint probability distribution to the be invariant under time shifts (stronger condition, rarely checked in practice). Most forcasting models (like ARIMA) assume or require stationarity because non stationary series ar harder to model and predict reliably future behavior depends on when you observe it.

** Testing for Stationarity**:
Common statistical tests include ⁉

- Augmented Dickey -Fulller (ADF) Test : Tests the null hypothesis that the series has a unit root (non-stationary). if p-value < 0.05 (or chosen significance level ), reject null - series is stationary.

- Kwiatkowski-Phillis-Schmidt-Shin(KPSS) Test- Tests the null hypothesis that the series is stationary. If p-value < 00.5, reject null - non- stationary.

- Visual methods : Plot the series, rolling mean/variance , or ACF plot (slow decay in ACF indicates non-stationarity).

**Transfoming a Non-Stationary Series into a Stationary One:**

1. Differencing : Subtract consecutive observations (first order : Y_t -Y_ {t-1}; seasonal differencing at lag m for seasonality). Removes trend and stabilizers mean. Often sufficient (d or D in ARIMA).
EX- For a series with upward trend, first differncing makes it fluctuate around zero.

2. Log Transformation: Apply natural log (or box-cox) to stabilize variance if it increases with the level (heteroscedasticity).
Ex- Stock prices or exponential growth data often become more stable after log.

3. Detrending: Subtract a fitted trend line (linear/polynomial regression ) from the series.

4. Seasonal Adjustment: Use seasonal differencing or decomposition methods (STL, X -13 ARIMA-SEATS) to remove seasonal effects.

5. Power Transformations(e.g., BOX-COX or Yeo-Johnson): Handle non-constant variance.



5. Q. Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in
terms of structure and application.

>> A. **Modeln > Structure (key Parameters) > Handles What? > When to Use**

- AR > Autoregressive: Current value depends on P past values (lags). Y_t = c + oY _{t-1} + ...+ o_PY_{t-p} + e-t > Short-term dependencies from past values > Simple stationary series with autocorrelation decaying gradually (e.g., temperature residuls). AR(p) is ARIMA (p,0,0).

- MR > Moving Average: Current value depends on q past errors (shocks). Y_t = c + e_t + 01e_{t-1} +..+ 0_q e_{t-q}  > Short terms shocks/noise effects > Stationary series where ACF cuts off after lag q (e.g., sudden impulse responses).  MA (q) is ARIMA (0,0,q).

- ARIMA > Autoregressive integrated moving average: Combines AR(p) + differencing(d) + MA(q).
ARIMA (p,d,q) on differenced series > Non-seasonal ,non-stationary series (trend via d) > General univeriate forecasting without seasonality (e.g., stock prices, sales without clear seasons). Requires stationarity after differencing.

- SARIMA > Seasonal ARIMA : Adds seaonal terms- (P, D,Q)m to ARIMA(p,d,q). Full: (p,d,q) (P,D,Q)m where m = seasonality period (e.g., 12 for monthly). > Non-stationary + seasonal patterns > Data with both trend and strong seasonality (e.g., montly airline passengers, retail sales, electricity demand ). Most common extension of ARIMA.

- SARIMAX > SARIMA with eXogenous variables: Adds external regressons (X) . Same as SARIMA but includes additional predictors  > seasonality + external factors influencing series > When external variables affect the target (e.g., forecasting the target(e.g., forecasting sales using advertising spend, holidays, weather, economic indicators as X) . Most flexible for multivariate influence.

**Summary of Differences:**

- AR and MA are basic building blocks (special cases of ARMA/ARIMA with d=0).

- ARIMA handles non-seasonal non- stationarity via integration (differencing).

- SARIMA extends ARIMA to capture repeating seasonal patterns.

- SARIMAX further extends SARIMA by incorporating exogenous (external) variables for better accuracy when other factors influence the series.

6. Q.  Load a time series dataset (e.g., AirPassengers), plot the original series,
and decompose it into trend, seasonality, and residual components
(Include your Python code and output in the code box below.)

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Load the dataset
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
data = pd.read_csv(url, parse_dates=['Month'], index_col='Month')
data = data.rename(columns={'Passengers': '#Passengers'})

# Plot original series
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['#Passengers'])
plt.title('Original AirPassengers Time Series')
plt.xlabel('Month')
plt.ylabel('Number of Passengers')
plt.grid(True)
plt.show()

# Decomposition (Multiplicative model)
decomposition = seasonal_decompose(data['#Passengers'], model='multiplicative', period=12)

# Plot decomposition
decomposition.plot()
plt.suptitle('AirPassengers Time Series Decomposition', fontsize=16)
plt.show()
```

Q.7 Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to
detect anomalies. Visualize the anomalies on a 2D scatter plot.
(Include your Python code and output in the code box below.)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Using a simple 2D synthetic dataset for clear visualization (you can replace with NYC Taxi data)
np.random.seed(42)
n_samples = 300

# Normal data
X_normal = np.random.normal(0, 1, (n_samples, 2))

# Add some anomalies
X_anomalies = np.random.uniform(-6, 6, (20, 2))
X = np.vstack([X_normal, X_anomalies])

# Fit Isolation Forest
iso_forest = IsolationForest(contamination=0.1, random_state=42)
y_pred = iso_forest.fit_predict(X)

# Visualization
plt.figure(figsize=(10, 6))
colors = np.where(y_pred == -1, 'red', 'blue')
plt.scatter(X[:, 0], X[:, 1], c=colors, alpha=0.7)
plt.title('Anomaly Detection using Isolation Forest')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend(['Normal', 'Anomaly'])
plt.grid(True)
plt.show()

print(f"Number of anomalies detected: {np.sum(y_pred == -1)}")
```

Q. 8 Train a SARIMA model on the monthly airline passengers dataset.
Forecast the next 12 months and visualize the results.
(Include your Python code and output in the code box below.)

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
import warnings
warnings.filterwarnings('ignore')

# Load data
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
data = pd.read_csv(url, parse_dates=['Month'], index_col='Month')

# Train SARIMA model (order chosen after analysis)
model = SARIMAX(data['Passengers'],
                order=(1, 1, 1),
                seasonal_order=(1, 1, 1, 12))

results = model.fit(disp=False)

# Forecast next 12 months
forecast = results.get_forecast(steps=12)
forecast_mean = forecast.predicted_mean
forecast_ci = forecast.conf_int()

# Plot
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Passengers'], label='Observed')
plt.plot(forecast_mean.index, forecast_mean, label='Forecast', color='red')
plt.fill_between(forecast_ci.index, forecast_ci.iloc[:, 0], forecast_ci.iloc[:, 1], color='pink', alpha=0.3)
plt.title('AirPassengers SARIMA Forecast (Next 12 Months)')
plt.xlabel('Month')
plt.ylabel('Passengers')
plt.legend()
plt.grid(True)
plt.show()

print("Forecast for next 12 months:")
print(forecast_mean)
```

Q. 9 Apply Local Outlier Factor (LOF) on any numerical dataset to detect
anomalies and visualize them using matplotlib.
(Include your Python code and output in the code box below.)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor
from sklearn.preprocessing import StandardScaler

# Create synthetic 2D dataset
np.random.seed(42)
n_samples = 300

# Normal data (two clusters)
X_normal1 = np.random.normal(0, 1, (150, 2))
X_normal2 = np.random.normal(4, 1.2, (150, 2))
X_normal = np.vstack([X_normal1, X_normal2])

# Add anomalies
X_anomalies = np.random.uniform(-5, 8, (25, 2))
X = np.vstack([X_normal, X_anomalies])

# Apply LOF
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.08)
y_pred = lof.fit_predict(X)
scores = lof.negative_outlier_factor_

# Visualization
plt.figure(figsize=(10, 6))
colors = np.where(y_pred == -1, 'red', 'blue')
plt.scatter(X[:, 0], X[:, 1], c=colors, alpha=0.7, edgecolors='k')
plt.title('Anomaly Detection using Local Outlier Factor (LOF)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend(['Normal', 'Anomaly'])
plt.grid(True)
plt.show()

print(f"Number of anomalies detected: {np.sum(y_pred == -1)}")
print(f"LOF scores (lower = more anomalous): {scores[:5]}")
```
